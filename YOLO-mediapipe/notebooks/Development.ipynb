# YOLO + MediaPipe Hand Tracking — Quickstart

Minimal example running the pipeline on a single static image, without opening a camera window — useful for quick checks in a notebook.


```python
import sys
sys.path.insert(0, "..")  # run from notebooks/ but import the repo-root src package

import cv2

from src.detection.detector import YOLODetector
from src.tracking.mediapipe_tracker import MediaPipeTracker
from src.tracking.tracker import HandTracker
from src.geometry.angle_calculator import AngleCalculator
from src.visualization.visualization import Visualizer
from src.utils.config import cfg
```

```python
detector = YOLODetector(cfg.yolo_model, confidence=cfg.confidence, device=cfg.device)
landmark_tracker = MediaPipeTracker(max_num_hands=cfg.max_hands)
tracker = HandTracker(detector, landmark_tracker, AngleCalculator())
visualizer = Visualizer()
```

```python
# Replace with a path to a real image containing a hand.
frame = cv2.imread("../assets/images/sample.jpg")

result = tracker.process(frame)
output = visualizer.draw_pipeline(
    frame=result.frame,
    detections=result.detections,
    hands=result.hands,
    poses=[],
    angles=result.angles,
    fps=result.fps,
)

print(result.angles)
```

```python
landmark_tracker.close()
```
